# Buliding my own LinearRegression Model.

## Import library 

```python
import numpy as np 
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler ,OneHotEncoder
```

## data preprocessing.

```python
df = pd.read_csv(r"C:\AI_ML\projects\CarsData.csv")
df.head()
```

```python
df.drop(columns='model',inplace=True)
df
```

```python
x_df = df.drop(['price'],axis=1)
target = df['price']
```

```python
ohe = OneHotEncoder(handle_unknown='ignore',sparse_output= False)

X_encoded = ohe.fit_transform(x_df[['transmission','fuelType','Manufacturer']])

col_names = ohe.get_feature_names_out(['transmission','fuelType','Manufacturer'])


df_encoded = pd.DataFrame(X_encoded,columns=col_names)


x_numeric = x_df.drop(columns=['transmission','fuelType','Manufacturer'])

final_df = pd.concat([df_encoded, x_numeric.reset_index(drop=True)], axis=1)

x_final = final_df
```

```python
std = StandardScaler()
data_std = std.fit_transform(x_final)
```

```python
x_std = pd.DataFrame(data_std,columns=std.get_feature_names_out())
x_std
```

```python
x = x_std
y = target
```

## Spliting the data (70 % fot traaining and 30 % for testing )

```python
x_train,x_test,y_train,y_test = train_test_split(x, y, test_size= 0.3, random_state=42)
```

## Model building

```python
class LinearRegression:

    def __init__(self,l_t = 1e-1):
        self.w = None
        self.b = 0
        self.l_t = l_t

    # Define gradient.    
    def gradient(self,X,Y):

        y_mean = np.mean(Y)

        prev_cost = float('inf')

        for i in range(5000):

            y_pred = np.dot(X , self.w) + self.b
            
            curr_cost = self.cost(Y , y_pred)

            dw = -(1/self.m) *np.dot(X.T , (Y - y_pred))
            db = -(1/self.m) *np.sum((Y - y_pred))

            self.w = self.w - self.l_t * dw
            self.b = self.b - self.l_t * db

            # To break iteration.
            if (prev_cost - curr_cost) < 1e-6:
                break

            prev_cost = curr_cost

            r2 = 1 - ((np.sum((Y - y_pred) **2 )) / (np.sum((Y - y_mean) **2 )))
            score = 100 * r2

            if i % 10 == 0:
                print(f"iter = {i}, score = {score}, cost= {curr_cost}")
        return score


    # Define a cost fucntion (Root mean square error)
    def cost(self,y_ture,y_pred):
        N = len(y_ture)
        loss = 1/N * np.sqrt(np.sum((y_ture - y_pred) **2 ))
        return loss
    
    
    def fit(self,x,y):

        self.m , self.n = x.shape
        self.w = np.zeros(self.n)

        score =  self.gradient(x,y)

        return score
    
    def score(self , x ,y):
        y_pred = np.dot(x,self.w) + self.b
        y_mean = np.mean(y)
        r2 = 1 - ((np.sum((y - y_pred) **2 )) / (np.sum((y - y_mean) **2 )))
        score = 100 * r2
        return score

    
    def predict(self,x):
        return np.dot(x,self.w) + self.b
```

```python
model = LinearRegression(1e-2)
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```
