# Data preprosessing 

```python
from sklearn.preprocessing import StandardScaler , OneHotEncoder 
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split,cross_val_score,cross_validate , KFold ,GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier 
```

```python
df_copy = df.copy()
df_copy.head()
```

```python
categorical_column = ['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope']
numric_columns = df_copy.drop(columns=['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope','HeartDisease'] ,axis=1).columns.tolist()
```

```python
numric_columns
```

```python
categorical_column
```

## Numric Data Pipeline.

```python
Numric_pipeline = Pipeline([
    ("SimpleImputer",SimpleImputer(strategy='median')),
    ("Scaler",StandardScaler())
])
```

## Categorical Data pipeline

```python
categorical_pipeline =Pipeline([
    ("ohe",OneHotEncoder(handle_unknown='ignore',sparse_output=False))
])
```

## Full Pipeline

```python
complete_pipeline = ColumnTransformer([
    ("num",Numric_pipeline,numric_columns),
    ("cat",categorical_pipeline,categorical_column)
])
```

```python
target = df['HeartDisease']
final_df = df_copy.drop(['HeartDisease'],axis=1)
```

```python
process_data = complete_pipeline.fit_transform(final_df)
process_data
```

# Model Training

```python
scoring = ['precision_macro', 'recall_macro','f1']
k = 10
kf = KFold(n_splits=k , shuffle=True , random_state=42)
```

#### Train a model using LogisticRegression 

```python
log_model = LogisticRegression()
log_scores = cross_validate(log_model,process_data,target,scoring=scoring,cv=kf)
# print(f"F1 scores of logisticReression : {log_model_f1}")
```

```python
log_scores.keys()
```

```python
print(f"F1 scores of logisticReression : \n{log_scores['test_f1']}\n mean : {log_scores['test_f1'].mean()}\n")
print(f"Precision scores of logisticReression : \n{log_scores['test_precision_macro']}\n mean : {log_scores['test_precision_macro'].mean()}\n")
print(f"Recall scores of logisticReression : \n{log_scores['test_recall_macro']}\n mean : {log_scores['test_recall_macro'].mean()}")
```

```python
plt.figure(figsize=(7,5))
plt.plot(range(10),log_scores['test_f1'],linestyle=":",label=['F1-score'],)
plt.plot(range(10),log_scores['test_precision_macro'],linestyle="-.",label=['Precision'])
plt.plot(range(10),log_scores['test_recall_macro'],label=["Recall"])
plt.legend()
plt.show()
```

```python
grid_param = {
    'C': [0.01, 0.1, 1, 10]
}

grid_scoring = {
    'precision': 'precision_macro',
    'recall': 'recall_macro',
    'f1': 'f1_macro'
}

log_grid_search = GridSearchCV(
    estimator=log_model,
    param_grid=grid_param,
    scoring=scoring,
    refit='f1',
    cv=kf
)
log_grid_search.fit(process_data,target)
```

```python
print(f"best parameter is : {log_grid_search.best_params_}")
print(f"best Score : {log_grid_search.best_score_}")
```

```python
log_grid_search.cv_results_
```

```python

result_df = pd.DataFrame(log_grid_search.cv_results_)

final_result = result_df[
    [
        'param_C',
        'mean_test_precision_macro',
        'mean_test_recall_macro',
        'mean_test_f1'
    ]
]

print(final_result)
```

```python
plt.figure(figsize=(8,7))
plt.plot(final_result['param_C'],final_result['mean_test_precision_macro'],label=['Precision'])
plt.plot(final_result['param_C'],final_result['mean_test_recall_macro'],label=['Recall'])
plt.plot(final_result['param_C'],final_result['mean_test_f1'],label=['F1-score'])
plt.legend()
plt.show()
```

#### Train a model using DecisionTree

```python
Tree_model = DecisionTreeClassifier(max_depth=20, min_samples_leaf=5, random_state= 42)
tree_model_scores = cross_validate(Tree_model,process_data,target,scoring=scoring,cv=kf)
```

```python
tree_model_scores.keys()
```

```python
print(f"F1 scores of DecisionTree : \n{tree_model_scores['test_f1']}\n mean : {tree_model_scores['test_f1'].mean()}\n")
print(f"Precision scores of DecisionTree : \n{tree_model_scores['test_precision_macro']}\nmean : {tree_model_scores['test_precision_macro'].mean()}\n")
print(f"Recall scores of DecisionTree : \n{tree_model_scores['test_recall_macro']}\nmean : {tree_model_scores['test_recall_macro'].mean()}\n")
```

```python
plt.figure(figsize=(7,5))
plt.plot(range(10),tree_model_scores['test_f1'],linestyle=":",label=['F1-score'],)
plt.plot(range(10),tree_model_scores['test_precision_macro'],linestyle="-.",label=['Precision'])
plt.plot(range(10),tree_model_scores['test_recall_macro'],label=["Recall"])
plt.legend()
plt.show()
```

#### Random Forest Model

```python
rf = RandomForestClassifier(random_state=42)

param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'max_features': ['sqrt']
}

grid = GridSearchCV(
    rf,
    param_grid,
    scoring='f1',
    cv=kf,
    n_jobs=-1
)

grid.fit(process_data, target)
```

```python
print(f"Best parameter : {grid.best_params_}")
print(f"Best score : {grid.best_score_}")
```
